```python
import numpy as np
import pandas as pd

CLEAN_PATH = "../data/dataset_delivery_time_clean.csv"

df = pd.read_csv(CLEAN_PATH)

lat1 = np.radians(df['Restaurant_latitude'])
lon1 = np.radians(df['Restaurant_longitude'])

lat2 = np.radians(df['Delivery_location_latitude'])
lon2 = np.radians(df['Delivery_location_longitude'])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
)

c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

R = 6371

df['Distance_km'] = R * c

df = df.drop(columns=['Restaurant_latitude', 'Restaurant_longitude', 'Delivery_location_latitude', 'Delivery_location_longitude'])

df["Order_hour"] = pd.to_datetime(df["Time_Order_picked"], format="%H:%M:%S").dt.hour

df.to_csv(CLEAN_PATH, index=False)
```
